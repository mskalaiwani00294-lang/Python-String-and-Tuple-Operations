# 2. String Slicing and Indexing

```python
#existing conccatenated string
text="Hello Zara,welcome to Python programming"
```

```python
#a.First character
print("a.First character:",text[0])
```

```python
#b.last character
print("b.last character:",text[-1])
```

```python
#c.firts 5 characters
print("c.firts 5 characters:",text[:6])
```

```python
#d.last 11 characters
print("d.last 11 characters:",text[-11:])
```

```python
#e.reverse the string
print("e.reverse the string:",text[::-1])
```

```python
#f.Print the word "Python"
print("f word Python:",text[22:28])
```

# 3. String Methods:

```python
strm="Python beginner tutorial"
```

```python
#a.covert the sentence to uppercase
print("Uppercase:",strm.upper())
```

```python
#b.convert the sentence to lowercase
print("lowercase:",strm.lower())
```

```python
#c.capitalize the sentence
print("capitalize:",strm.capitalize())
```

```python
#d.Count the total number of occurrences of character"t"
print("count of t:",strm.count("t"))
```

```python
#e Replace"Python" with" Data Analytics"
print ("After Replacement:",strm.replace("Python"," Data Analytics"))
```

# Tuples (Creation, Modification and Access)

```python
#Tuples- Creation, Modification and Access
#Create two tuples
t1=(10,20,30)
t2=(40,50,60)
```

```python
#a.concatenate the two tuples
t_combine=t1+t2
print("Combined Tuple:",t_combine)
```

```python
#b.Repeat the element of t_combine 3 times
print("Repeated Tuple:",t_combine*3)
```

```python
#c.Access the 3rd element from t_combine
print("3rd element:",t_combine[2])
```

```python
#d Access the first three elements
print("First 3 elements:",t_combine[:3])
```

```python
#e. last three elements
print("last three elements:",t_combine[-3:])
```
